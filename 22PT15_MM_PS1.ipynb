Analyze the dynamical system an+1= rna0 with a0=0 with

i) r=0

ii) 0<r<1

iii) -1<r<0

iv) r>1


```python
import matplotlib.pyplot as plt

# Initial value
a0 = 0.5  # (or) a0=0

# Number of iterations
N = 20

# Values of r for different cases
r_values = [0, 0.5, -0.5, 1.2]

titles = [
    "r = 0",
    "0 < r < 1 (r = 0.5)",
    "-1 < r < 0 (r = -0.5)",
    "|r| > 1 (r = 1.2)"
]

plt.figure(figsize=(12,8))

for i in range(len(r_values)):
    r = r_values[i]

    a = [a0]

    for _ in range(N):
        a.append(r * a[-1])

    plt.subplot(2,2,i+1)
    plt.plot(range(N+1), a, marker='o')
    plt.title(titles[i])
    plt.xlabel("Iteration")
    plt.ylabel("a(n)")
    plt.grid(True)

plt.tight_layout()
plt.show()
```

2. Construct a dynamical system for the following : Consider the decay of digoxin in the blood stream to precribed dosage that keeps the concentration between the acceptable levels. Suppose we prescribe a daily dosage of (i)  0.1 mg (ii) 0.2 mg (iii) 0.3 mg and know that half the digoxin remains in the system at the end of each dosage period. Plot he graphs and discuss the stability.


```python
import matplotlib.pyplot as plt

'''
# modelling

an = concentration after the nthday
d = daily dosage

remaining medicine by eod -> 0.5*n
daily dose -> d

a_n+1 = 0.5*a_n + d
'''

# Daily dosages
dosages = [0.1, 0.2, 0.3]

days = 20

plt.figure(figsize=(10, 6))

for dose in dosages:
    concentration = [0]  # initially 0

    for _ in range(days):
        next_value = 0.5 * concentration[-1] + dose
        concentration.append(next_value)
    plt.plot(range(days + 1), concentration, marker='o',
             label=f'Dose = {dose} mg')

plt.xlabel("Day")
plt.ylabel("Digoxin Concentration (mg)")
plt.title("Digoxin Concentration in Bloodstream")
plt.grid(True)
plt.legend()
plt.show()
```

3. Generate (i) 500, (ii) 1000, (iii) 10,000 and (iv) 1,00,000 random numbers for the following distributions. Draw the histogram and frequency polygon
(i) Uniform distribution   (ii) exponential distribution   (iii) Weibull distribution  (iv) triangular distribution


```python
import numpy as np
import matplotlib.pyplot as plt


# Generate random numbers
sample_sizes = [500, 1000, 10000, 100000]

for N in sample_sizes:
    print(f"\nGenerating {N} random numbers...")

    uniform = np.random.uniform(0, 1, N)
    exponential = np.random.exponential(scale=1, size=N)
    weibull = np.random.weibull(a=2, size=N)
    triangular = np.random.triangular(left=0, mode=5, right=10, size=N)

    distributions = [
        ("Uniform", uniform),
        ("Exponential", exponential),
        ("Weibull", weibull),
        ("Triangular", triangular)
    ]

    plt.figure(figsize=(12, 10))

    for i, (name, data) in enumerate(distributions):

        plt.subplot(2, 2, i + 1)

        # Histogram
        counts, bins, _ = plt.hist(
            data,
            bins=20,
            alpha=0.6,
            edgecolor='black',
            label='Histogram'
        )

        # Frequency Polygon
        centers = (bins[:-1] + bins[1:]) / 2
        plt.plot(centers, counts, marker='o', label='Frequency Polygon')

        plt.title(name)
        plt.xlabel("Value")
        plt.ylabel("Frequency")
        plt.legend()

    plt.tight_layout()
    plt.show()
```

4. Obtain the value of pie using Monte carlo simulation method by generating random numbers by between [0,1]  using the random number generated in problem (1)


```python
import numpy as np
import matplotlib.pyplot as plt

# Number of random points
N = 100000

# Generate random x and y coordinates
x = np.random.uniform(0, 1, N)
y = np.random.uniform(0, 1, N)

# Check whether each point lies inside the quarter circle
inside = (x**2 + y**2) <= 1

# Count the number of points inside
inside_count = np.sum(inside)

# Estimate pi
pi_estimate = 4 * inside_count / N

print("Total Points :", N)
print("Points Inside Quarter Circle :", inside_count)
print("Estimated Pi :", pi_estimate)
print("Actual Pi :", np.pi)
print("Absolute Error :", abs(np.pi - pi_estimate))

# visualization
plt.figure(figsize=(6,6))

plt.scatter(x[inside], y[inside], s=5, label="Inside Circle")
plt.scatter(x[~inside], y[~inside], s=5, label="Outside Circle")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Monte Carlo Estimation of Pi")
plt.axis("equal")
plt.legend()
plt.grid(True)
plt.show()
```

  6. The inter arrival time of the customer at a booking station is exponential with men 10 min and the service time for booking follows Poisson distribution between 8 and 12 min. The customers are served FIFO basis. Simulate the system for 1000 time units . From the results of the simulation determine the average time customer waits in a queue, average number of customers waiting and average utilization of the booking station Plot the sample path respect to the time.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Simulation parameters
SIMULATION_TIME = 1000

arrival_times = []
service_times = []

current_time = 0

# Generate customers until simulation time ends
while current_time < SIMULATION_TIME:
    interarrival = np.random.exponential(scale=10)
    current_time += interarrival

    if current_time > SIMULATION_TIME:
        break

    arrival_times.append(current_time)

    # Poisson service time (restricted to 8–12 minutes)
    service = np.random.poisson(lam=10)
    service = max(8, min(12, service))

    service_times.append(service)

n = len(arrival_times)

start_times = []
finish_times = []
waiting_times = []

for i in range(n):
    if i == 0:
        start = arrival_times[i]
    else:
        start = max(arrival_times[i], finish_times[i-1])

    finish = start + service_times[i]
    wait = start - arrival_times[i]

    start_times.append(start)
    finish_times.append(finish)
    waiting_times.append(wait)

# Create table
df = pd.DataFrame({
    "Arrival": arrival_times,
    "Service": service_times,
    "Start": start_times,
    "Finish": finish_times,
    "Waiting": waiting_times
})

print(df)

# Performance Measures
avg_wait = np.mean(waiting_times)
busy_time = sum(service_times)
utilization = busy_time / SIMULATION_TIME
avg_queue = sum(waiting_times) / SIMULATION_TIME

print("\nAverage Waiting Time :", avg_wait)
print("Average Queue Length :", avg_queue)
print("Server Utilization :", utilization)

# Sample Path
plt.figure(figsize=(10,5))
plt.step(range(n), waiting_times, where='mid')
plt.xlabel("Customer")
plt.ylabel("Waiting Time")
plt.title("Sample Path of Waiting Time")
plt.grid(True)
plt.show()
```

5. Simulate the following problem for 200, 500, 1000 and 10,000 days.

A newsstand buys newspapers for 30 cents each and sells them for 45 cents each. Newspapers that are not sold by the end of the day are sold as scrap for 5 cents each.

News types and their probabilities:

| News Type   | Probability | Cumulative Probability | Random Digit |
| ----------- | ----------: | ---------------------: | ------------ |
| Good        |        0.35 |                   0.35 | 01–35        |
| Fair/Normal |        0.45 |                   0.80 | 36–80        |
| Poor        |        0.20 |                   1.00 | 81–00        |

a) Generate 100 random numbers using Uniform Distribution (0,100) and determine the type of news.

b) Generate demand values between 0 and 100 using:

* Good news → Exponential Distribution, mean = 50
* Fair/Normal news → Normal Distribution, mean = 50, SD = 10
* Poor news → Poisson Distribution, mean = 50

Display all generated numbers.

Demand distribution:

| Demand | Good CDF | Fair CDF | Poor CDF | Good  | Fair  | Poor  |
| -----: | -------: | -------: | -------: | ----- | ----- | ----- |
|     40 |     0.03 |     0.10 |     0.44 | 01–03 | 01–10 | 01–44 |
|     50 |     0.08 |     0.28 |     0.66 | 04–08 | 11–28 | 45–66 |
|     60 |     0.23 |     0.68 |     0.82 | 09–23 | 29–68 | 67–82 |
|     70 |     0.43 |     0.88 |     0.94 | 24–43 | 69–88 | 83–94 |
|     80 |     0.78 |     0.96 |     1.00 | 44–78 | 89–96 | 95–00 |
|     90 |     0.93 |     1.00 |     1.00 | 79–93 | 97–00 | -     |
|    100 |     1.00 |     1.00 |     1.00 | 94–00 | -     | -     |

For each day, calculate:

1. Revenue from sales
   Revenue = Newspapers Sold × 0.45

2. Loss/cost due to excess newspapers
   Purchase Cost = Newspapers Purchased × 0.30

3. Salvage from unsold newspapers
   Salvage = Unsold Newspapers × 0.05

4. Daily Profit
   Daily Profit = Revenue + Salvage − Purchase Cost

where:

Newspapers Sold = min(Newspapers Purchased, Demand)

Unsold Newspapers = max(Newspapers Purchased − Demand, 0)

For each simulation period (200, 500, 1000 and 10,000 days), display the results and calculate the average daily profit.


```python
import numpy as np
import pandas as pd

# Prices (cents)
BUY_PRICE = 30
SELL_PRICE = 45
SCRAP_PRICE = 5

# Number of newspapers purchased daily
INVENTORY = 70

# For reproducibility
np.random.seed(10)


# Convert generated random value (0-100) to demand
def get_demand(news_type, rn):
    rn = int(round(rn))
    rn = max(1, min(100, rn))  # Restrict between 1 and 100

    if news_type == "Good":
        if rn <= 3:
            return 40
        elif rn <= 8:
            return 50
        elif rn <= 23:
            return 60
        elif rn <= 43:
            return 70
        elif rn <= 78:
            return 80
        elif rn <= 93:
            return 90
        else:
            return 100

    elif news_type == "Fair":
        if rn <= 10:
            return 40
        elif rn <= 28:
            return 50
        elif rn <= 68:
            return 60
        elif rn <= 88:
            return 70
        elif rn <= 96:
            return 80
        else:
            return 90

    else:   # Poor

        if rn <= 44:
            return 40
        elif rn <= 66:
            return 50
        elif rn <= 82:
            return 60
        elif rn <= 94:
            return 70
        else:
            return 80



# Simulation
for DAYS in [200, 500, 1000, 10000]:

    print("\n")
    print("=" * 80)
    print(f"SIMULATION FOR {DAYS} DAYS")
    print("=" * 80)

    results = []

    for day in range(1, DAYS + 1):
        # Step 1 : News Type
        news_rn = np.random.randint(1, 101)

        if news_rn <= 35:
            news_type = "Good"
            generated = np.random.exponential(scale=50)

        elif news_rn <= 80:
            news_type = "Fair"
            generated = np.random.normal(loc=50, scale=10)

        else:
            news_type = "Poor"
            generated = np.random.poisson(lam=50)

        # Restrict generated number between 0 and 100
        generated = max(0, min(100, generated))


        # Step 2 : Demand
        demand = get_demand(news_type, generated)

        sold = min(demand, INVENTORY)
        unsold = INVENTORY - sold
        revenue = sold * SELL_PRICE
        scrap = unsold * SCRAP_PRICE
        purchase_cost = INVENTORY * BUY_PRICE
        daily_profit = revenue + scrap - purchase_cost
        lost_profit = max(0, demand - INVENTORY) * (SELL_PRICE - BUY_PRICE)

        results.append([
            day,
            news_rn,
            news_type,
            round(generated, 2),
            demand,
            sold,
            unsold,
            revenue,
            scrap,
            lost_profit,
            daily_profit
        ])

    df = pd.DataFrame(results,
                      columns=[
                          "Day",
                          "News RN",
                          "News Type",
                          "Generated RN",
                          "Demand",
                          "Sold",
                          "Unsold",
                          "Revenue",
                          "Scrap Value",
                          "Lost Profit",
                          "Daily Profit"
                      ])

    print(df)

    print("\nSUMMARY")
    print("-" * 40)

    print("Total Revenue      :", df["Revenue"].sum())
    print("Total Scrap Value  :", df["Scrap Value"].sum())
    print("Total Lost Profit  :", df["Lost Profit"].sum())
    print("Total Profit       :", df["Daily Profit"].sum())
    print("Average Profit/Day :", round(df["Daily Profit"].mean(), 2))
```

```python
import numpy as np
import pandas as pd
import random

# Constants
BUY_PRICE = 30
SELL_PRICE = 45
SCRAP_PRICE = 5
PURCHASED_QTY = 70  # Assumed stock quantity

def get_news_type(rn):
    """Determine news type based on random number 1-100"""
    if 1 <= rn <= 35: return "Good"
    elif 36 <= rn <= 80: return "Fair"
    else: return "Poor"

def get_demand(rn, news_type):
    """Determine demand based on news type and random number 1-100"""
    # Demand Tables mapping (Cumulative Probability Range : Demand Value)
    if news_type == "Good":
        if rn <= 3: return 40
        elif rn <= 8: return 50
        elif rn <= 23: return 60
        elif rn <= 43: return 70
        elif rn <= 78: return 80
        elif rn <= 93: return 90
        else: return 100
    elif news_type == "Fair":
        if rn <= 10: return 40
        elif rn <= 28: return 50
        elif rn <= 68: return 60
        elif rn <= 88: return 70
        elif rn <= 96: return 80
        else: return 90 # Max for fair is 90 according to table (97-00)
    else: # Poor
        if rn <= 44: return 40
        elif rn <= 66: return 50
        elif rn <= 82: return 60
        elif rn <= 94: return 70
        else: return 80 # Max for poor is 80 according to table (95-00)

def simulate(days):
    total_revenue = 0
    total_lost_profit = 0
    total_salvage = 0
    total_profit = 0

    for _ in range(days):
        # Generate random digits for type and demand
        rn_type = random.randint(1, 100)
        rn_demand = random.randint(1, 100)

        type_of_news = get_news_type(rn_type)
        demand = get_demand(rn_demand, type_of_news)

        # Calculations
        sold = min(demand, PURCHASED_QTY)
        excess = max(0, demand - PURCHASED_QTY)
        unsold = max(0, PURCHASED_QTY - demand)

        revenue = sold * SELL_PRICE
        salvage = unsold * SCRAP_PRICE
        loss_of_profit = excess * (SELL_PRICE - BUY_PRICE)
        daily_cost = PURCHASED_QTY * BUY_PRICE
        profit = revenue + salvage - daily_cost

        # Accumulate
        total_revenue += revenue
        total_lost_profit += loss_of_profit
        total_salvage += salvage
        total_profit += profit

    return {
        "Days": days,
        "Total Revenue": total_revenue,
        "Total Lost Profit": total_lost_profit,
        "Total Salvage": total_salvage,
        "Total Profit": total_profit,
        "Avg Profit/Day": total_profit / days
    }

# Execution for different day counts
results = []
for d in [200, 500, 1000, 10000]:
    results.append(simulate(d))

# Part A & B: Random Number Generation Demonstration
print("--- Part A: 100 Random Numbers (Uniform) ---")
uniform_rns = [random.randint(1, 100) for _ in range(100)]
print(uniform_rns)

print("\n--- Part B: Special Distributions for Random Digits (0-100) ---")
# Good news: Exponential (mean 50) clipped to 100
good_dist = np.clip(np.random.exponential(50, 10), 0, 100).astype(int)
# Poor news: Poisson (mean 50) clipped to 100
poor_dist = np.clip(np.random.poisson(50, 10), 0, 100).astype(int)
# Fair news: Normal (mean 50, SD 10)
fair_dist = np.clip(np.random.normal(50, 10, 10), 0, 100).astype(int)

print(f"Good (Exp): {good_dist}")
print(f"Poor (Poiss): {poor_dist}")
print(f"Fair (Norm): {fair_dist}")

print("\n--- Simulation Results ---")
df = pd.DataFrame(results)
print(df.to_string(index=False))
```
